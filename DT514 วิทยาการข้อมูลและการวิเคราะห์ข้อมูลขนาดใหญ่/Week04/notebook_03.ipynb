<small>26. Check the data type of "Customer_Days" and then change its data type to an appropriate date-related data type and check the result again.

<small>28. Divide customers into three groups: (1) Income < 30,000 → Low, (2) 30,000–60,000 → Medium, (3) Income > 60,000 → High


```python
df["Income_Group"] = np.where(df["Income"] < 30000, "Low",
                      np.where(df["Income"] <= 60000, "Medium", "High"))

print(df[["Income", "Income_Group"]].head())
```

<small> 29. Group customers by "Education" and calculate: (1) Number of customers, (2) Average Income, (3) Average Age, (4) Average Wine Spending, (5) Average Web Purchases, (6) Average Store Purchases


```python
grouped = df.groupby("Education").agg({
    "No": "count",
    "Income": "mean",
    "Age": "mean",
    "MntWines": "mean",
    "NumWebPurchases": "mean",
    "NumStorePurchases": "mean"
}).rename(columns={
    "No": "Number_of_Customers",
    "Income": "Average_Income",
    "Age": "Average_Age",
    "MntWines": "Average_Wine_Spending",
    "NumWebPurchases": "Average_Web_Purchases",
    "NumStorePurchases": "Average_Store_Purchases"
})
print(grouped)
```

<small>30. Create a Pivot Table with: Index = Education and Columns = Response and Values = Number of customer


```python
pivot_table = pd.pivot_table(df,
                             index="Education",
                             columns="Response",
                             values="No",
                             aggfunc="count",
                             fill_value=0)

pivot_table
```

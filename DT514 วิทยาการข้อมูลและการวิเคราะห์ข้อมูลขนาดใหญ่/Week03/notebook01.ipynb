```python
import numpy as np
print(np.__version__)
```

```python
#arr = [1, 2, 3, 4, 5]
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
arange_arr = np.arange(1, 10, 2) # start = 1, stop = 10, step = 2 (default step = 1)
print(arange_arr)
```

```python
linspace_arr = np.linspace(1, 10, 5) # start = 1, stop = 10, num = 5 (default num = 50)
print(linspace_arr)
```

```python
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d)
```

```python
print("type:", arr_2d.dtype)
print("shape:", arr_2d.shape) #row, column
```

```python
# Example 1: Indexing
print("Element at [0,1]:", arr_2d[0, 1]) # row 0, column 1
print("Element at [1,2]:", arr_2d[1, 2]) # row 1, column 2

# Example 2: Slicing
print("First row:", arr_2d[0, :]) # all columns of row 0
print("Second column:", arr_2d[:, 1]) # all rows of column 1

# Example 3: Boolean Indexing
print("Elements greater than 3:", arr_2d[arr_2d > 3]) # elements greater than 3
print("Elements less than or equal to 3:", arr_2d[arr_2d <= 3]) # elements less than or equal to 3
print("Elements equal to 5:", arr_2d[arr_2d == 5]) # elements equal to 5
```

```python
# Reshaping arrays
reshaped_arr = arr_2d.reshape(3, 2) # reshape to 3 rows and 2 columns
print("Reshaped array (3x2):\n", reshaped_arr)

# raveling arrays
raveled_arr = arr_2d.ravel() # flatten the array
print("Raveled array:", raveled_arr)

# Flattening arrays
flattened_arr = arr_2d.flatten() # flatten the array
print("Flattened array:", flattened_arr)
```

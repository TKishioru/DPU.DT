Section A: Distance Metrics

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics.pairwise import (
    euclidean_distances,
    manhattan_distances,
    cosine_similarity,
    cosine_distances
)
from scipy.spatial.distance import minkowski


# 1. Create Sample Data
A = np.array([[1, 1]])
B = np.array([[4, 5]])


# 2. Calculate Distance Metrics
euclidean = euclidean_distances(A, B)[0][0]
manhattan = manhattan_distances(A, B)[0][0]
minkowski_dist = minkowski(A[0], B[0], p=3)
cosine_dist = cosine_distances(A, B)[0][0]
cosine_sim = cosine_similarity(A, B)[0][0]


# 3. Display Results
results = pd.DataFrame({
    'Metric': [
        'Euclidean Distance',
        'Manhattan Distance',
        'Minkowski Distance (p=3)',
        'Cosine Distance',
        'Cosine Similarity'
    ],
    'Value': [
        euclidean,
        manhattan,
        minkowski_dist,
        cosine_dist,
        cosine_sim
    ]
})

print(results.round(4))


# 4. Visualize Distance
plt.figure(figsize=(7, 6))

# Plot points
plt.scatter(A[0, 0], A[0, 1],
            color='blue', s=100, label='Point A')

plt.scatter(B[0, 0], B[0, 1],
            color='red', s=100, label='Point B')

# Euclidean Distance
plt.plot(
    [A[0, 0], B[0, 0]],
    [A[0, 1], B[0, 1]],
    'b--',
    label='Euclidean Distance'
)

# Manhattan Distance
plt.plot(
    [A[0, 0], B[0, 0], B[0, 0]],
    [A[0, 1], A[0, 1], B[0, 1]],
    'r-',
    label='Manhattan Distance'
)

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Euclidean vs Manhattan Distance')
plt.grid(True)
plt.legend()
plt.show()


# 5. Visualize Cosine Similarity
plt.figure(figsize=(7, 7))

# Vector A
plt.quiver(
    0, 0, A[0, 0], A[0, 1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label='Vector A'
)

# Vector B
plt.quiver(
    0, 0, B[0, 0], B[0, 1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='red',
    label='Vector B'
)

# Display angle and similarity
plt.text(
    0.5, 5.8,
    f'Cosine Similarity = {cosine_sim:.4f}\n'
    f'Angle = {angle:.2f} degrees',
    fontsize=11
)

# Plot settings
plt.xlim(-1, 6)
plt.ylim(-1, 7)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Cosine Similarity: Angle Between Vectors')

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal')
plt.show()
```

Section B: Example of K-means Clustering Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# 1. Generate Sample Data
X, _ = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=1.0,
    random_state=42
)

# 2. Elbow Method
k_values = range(1, 11)
inertia = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)
    inertia.append(model.inertia_)

# 3. Visualize Elbow Method
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    k_values,
    inertia,
    marker='o',
    markersize=7,
    linewidth=2
)

ax.set_xlabel('Number of Clusters (K)', fontsize=12)
ax.set_ylabel('Inertia (WCSS)', fontsize=12)
ax.set_title('Elbow Method for Optimal K',
             fontsize=14, fontweight='bold')

ax.set_xticks(list(k_values))
ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()

# 4. Apply K-Means Clustering
optimal_k = 4

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(X)

centroids = kmeans.cluster_centers_

# 5. Visualize Clusters
fig, ax = plt.subplots(figsize=(8, 6))

# Plot data points
scatter = ax.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap='viridis',
    s=40,
    alpha=0.7,
    edgecolors='white',
    linewidths=0.3
)

# Plot centroids
ax.scatter(
    centroids[:, 0],
    centroids[:, 1],
    s=250,
    c='red',
    marker='X',
    edgecolors='black',
    linewidths=1,
    label='Centroids',
    zorder=3
)

ax.set_xlabel('Feature 1', fontsize=12)
ax.set_ylabel('Feature 2', fontsize=12)
ax.set_title('K-Means Clustering',
             fontsize=14, fontweight='bold')

ax.legend()
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

Section C: Example of Heirachical Clustering

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# 1. Generate Sample Data
X, _ = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=1.0,
    random_state=42
)


# 2. Hierarchical Clustering

# Calculate linkage matrix
Z = linkage(
    X,
    method='complete',
    metric='euclidean'
)


# 3. Visualize Dendrogram

fig, ax = plt.subplots(figsize=(12, 6))

dendrogram(
    Z,
    truncate_mode='lastp',
    p=30,
    leaf_rotation=90,
    leaf_font_size=9,
    show_contracted=True,
    ax=ax
)

ax.set_xlabel('Data Points / Merged Clusters', fontsize=12)
ax.set_ylabel('Distance', fontsize=12)
ax.set_title('Hierarchical Clustering Dendrogram',
             fontsize=14, fontweight='bold')

ax.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()


# 4. Apply Agglomerative Clustering
optimal_k = 4

model = AgglomerativeClustering(
    n_clusters=optimal_k,
    metric='euclidean',
    linkage='complete'
)

labels = model.fit_predict(X)


# 5. Calculate Cluster Centroids
centroids = np.array([
    X[labels == i].mean(axis=0)
    for i in range(optimal_k)
])


# 6. Visualize Clusters
fig, ax = plt.subplots(figsize=(8, 6))

# Plot data points
scatter = ax.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap='viridis',
    s=40,
    alpha=0.7,
    edgecolors='white',
    linewidths=0.3
)

# Plot cluster centroids
ax.scatter(
    centroids[:, 0],
    centroids[:, 1],
    s=250,
    c='red',
    marker='X',
    edgecolors='black',
    linewidths=1,
    label='Cluster Centers',
    zorder=3
)

ax.set_xlabel('Feature 1', fontsize=12)
ax.set_ylabel('Feature 2', fontsize=12)
ax.set_title('Hierarchical Clustering',
             fontsize=14, fontweight='bold')

ax.legend()
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

Section D: Example of DBSCAN Clustering

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors

# 1. Generate Sample Data
X, _ = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=1.0,
    random_state=42
)


# 2. K-Distance Graph
min_samples = 5
nearest_neighbors = NearestNeighbors(
    n_neighbors = min_samples
)

nearest_neighbors.fit(X)
distances, indices = nearest_neighbors.kneighbors(X)

# Sort distances to the k-th nearest neighbor
k_distances = np.sort(distances[:, -1])


# 3. Visualize K-Distance Graph
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(
    np.arange(len(k_distances)),
    k_distances,
    linewidth=2
)
ax.set_xlabel('Points Sorted by Distance', fontsize=12)
ax.set_ylabel(f'{min_samples}-Nearest Neighbor Distance',
              fontsize=12)
ax.set_title('K-Distance Graph for Selecting eps',
             fontsize=14, fontweight='bold')
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()


# 4. Apply DBSCAN
eps = 0.8
dbscan = DBSCAN(
    eps=eps,
    min_samples=min_samples,
    metric='euclidean'
)

labels = dbscan.fit_predict(X)


# 5. Analyze Clustering Results
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = np.sum(labels == -1)

print(f"Number of Clusters: {n_clusters}")
print(f"Number of Noise Points: {n_noise}")


# 6. Identify Core Points and Noise
core_mask = np.zeros(len(X), dtype=bool)
core_mask[dbscan.core_sample_indices_] = True


# 7. Visualize DBSCAN Clustering
fig, ax = plt.subplots(figsize=(9, 7))

# Plot clustered data points
scatter = ax.scatter(
    X[~core_mask & (labels != -1), 0],
    X[~core_mask & (labels != -1), 1],
    c=labels[~core_mask & (labels != -1)],
    cmap='viridis',
    s=35,
    alpha=0.7,
    edgecolors='white',
    linewidths=0.3,
    label='Border Points'
)

# Plot core points
ax.scatter(
    X[core_mask, 0],
    X[core_mask, 1],
    c=labels[core_mask],
    cmap='viridis',
    s=65,
    edgecolors='black',
    linewidths=0.7,
    label='Core Points'
)

# Plot noise points
ax.scatter(
    X[labels == -1, 0],
    X[labels == -1, 1],
    c='red',
    marker='x',
    s=70,
    linewidths=2,
    label='Noise Points'
)

ax.set_xlabel('Feature 1', fontsize=12)
ax.set_ylabel('Feature 2', fontsize=12)

ax.set_title(
    f'DBSCAN Clustering (eps={eps}, min_samples={min_samples})',
    fontsize=14,
    fontweight='bold'
)

ax.legend()
ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

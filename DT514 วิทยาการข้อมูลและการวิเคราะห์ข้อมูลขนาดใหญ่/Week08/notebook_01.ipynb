```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Section 01: Linear Regression

<small> 2 Dimensions Model </small>

```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression

# 1. Generate mockup data
np.random.seed(42)
X = 2 * np.random.rand(200, 1)
y = 5 + 3 * X + np.random.randn(200, 1) * 0.8

# 2. Train Linear Regression
model = LinearRegression()
model.fit(X, y)

# 3. Create points for the regression line
X_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_line = model.predict(X_line)

# 4. Plot
plt.figure(figsize=(8, 5))

plt.scatter(X, y, alpha=0.6, label='Actual Data')
plt.plot(X_line, y_line, color='red', linewidth=2,
         label='Regression Line')

plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression (200 Samples)')
plt.legend()
plt.grid(True)
plt.show()

# 5. Display equation
print(
    f"Regression Equation: " 
    f"y = {model.coef_[0, 0]:.2f}x + {model.intercept_[0]:.2f}"
)
```

<small> 3 Dimensions Model </small>

```python
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression

# 1. Generate mockup data (200 samples)
np.random.seed(42)

X1 = 2 * np.random.rand(200, 1)
X2 = 3 * np.random.rand(200, 1)

X = np.hstack((X1, X2))

y = 5 + 3 * X1 + 2 * X2 + np.random.randn(200, 1) * 0.8

# 2. Train Multiple Linear Regression
model = LinearRegression()
model.fit(X, y)

# 3. Create points for the regression plane
x1_line, x2_line = np.meshgrid(
    np.linspace(X1.min(), X1.max(), 30),
    np.linspace(X2.min(), X2.max(), 30)
)

X_plane = np.column_stack((
    x1_line.ravel(),
    x2_line.ravel()
))

y_plane = model.predict(X_plane).reshape(x1_line.shape)

# 4. Create interactive 3D plot
fig = go.Figure()

# Actual data
fig.add_trace(go.Scatter3d(
    x=X1.flatten(),
    y=X2.flatten(),
    z=y.flatten(),
    mode='markers',
    name='Actual Data',
    marker=dict(
        size=4,
        color='blue',
        opacity=0.7
    )
))

# Regression plane
fig.add_trace(go.Surface(
    x=x1_line,
    y=x2_line,
    z=y_plane,
    name='Regression Plane',
    colorscale='Reds',
    opacity=0.5,
    showscale=False
))

# 5. Layout
fig.update_layout(
    title='Multiple Linear Regression (200 Samples)',
    scene=dict(
        xaxis_title='X1',
        yaxis_title='X2',
        zaxis_title='y'
    ),
    width=900,
    height=700
)

fig.show()

# 6. Display equation
print("Regression Equation:")
print(
    f"y = {model.coef_[0, 0]:.2f}X1 "
    f"+ {model.coef_[0, 1]:.2f}X2 "
    f"+ {model.intercept_[0]:.2f}"
)
```

Section 02: Polinomial Regression Degree 2

<small>2 Dimensions Model</small>

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error, r2_score

# 1. Generate Nonlinear Mockup Data (200 samples)
np.random.seed(42)
X = np.random.uniform(-3, 3, (200, 1))
y = 2 * X[:, 0]**2 + 3 * X[:, 0] + 5
y += np.random.normal(0, 3, 200)


# 2. Create and train Polynomial Regression model
model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

model.fit(X, y)

# 3. Generate points for the regression curve
X_line = np.linspace(-3, 3, 300).reshape(-1, 1)
y_line = model.predict(X_line)

# 4. Visualize data and regression curve
plt.figure(figsize=(8, 5))

plt.scatter(X, y, alpha=0.6, label='Actual Data')
plt.plot(
    X_line,
    y_line,
    color='red',
    linewidth=2,
    label='Regression Curve'
)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Polynomial Regression (200 Samples)')
plt.legend()
plt.grid(True)
plt.show()

# 5. Display the learned regression equation
poly = model.named_steps['polynomialfeatures']
lr = model.named_steps['linearregression']

print("Regression Equation:")
print(
    f"y = {lr.intercept_:.2f} "
    f"+ ({lr.coef_[1]:.2f})X "
    f"+ ({lr.coef_[2]:.2f})X^2"
)
```

<small>3 Dimensions Model</small>

```python
import plotly.graph_objects as go
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# 1. Generate Nonlinear Mockup Data (200 samples)
np.random.seed(42)

X1 = np.random.uniform(-3, 3, 200)
X2 = np.random.uniform(-3, 3, 200)

y = (
    2 * X1**2
    + 1.5 * X2**2
    + 3 * X1 * X2
    + 3 * X1
    + 2 * X2
    + 5
)

y += np.random.normal(0, 3, 200)

# Combine X1 and X2
X = np.column_stack((X1, X2))


# 2. Create and train Polynomial Regression model
model = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

model.fit(X, y)


# 3. Generate points for the regression surface
X1_line = np.linspace(-3, 3, 50)
X2_line = np.linspace(-3, 3, 50)

X1_grid, X2_grid = np.meshgrid(X1_line, X2_line)

X_grid = np.column_stack((
    X1_grid.ravel(),
    X2_grid.ravel()
))

y_grid = model.predict(X_grid)
y_grid = y_grid.reshape(X1_grid.shape)


# 4. Visualize data and regression surface
fig = go.Figure()

# Actual data points
fig.add_trace(go.Scatter3d(
    x=X1,
    y=X2,
    z=y,
    mode='markers',
    marker=dict(
        size=4,
        color='blue',
        opacity=0.7
    ),
    name='Actual Data'
))

# Polynomial regression surface
fig.add_trace(go.Surface(
    x=X1_grid,
    y=X2_grid,
    z=y_grid,
    colorscale='Reds',
    opacity=0.6,
    name='Regression Surface'
))

fig.update_layout(
    title='Polynomial Regression (Degree 2)',
    scene=dict(
        xaxis_title='X1',
        yaxis_title='X2',
        zaxis_title='y'
    ),
    width=900,
    height=700
)

fig.show()


# 5. Display the learned regression equation
poly = model.named_steps['polynomialfeatures']
lr = model.named_steps['linearregression']

feature_names = poly.get_feature_names_out(['X1', 'X2'])

print("\nPolynomial Regression Equation:")

equation = f"y = {lr.intercept_:.2f}"

for coef, name in zip(lr.coef_, feature_names):
    if name != '1':
        sign = '+' if coef >= 0 else '-'
        equation += f" {sign} {abs(coef):.2f}({name})"

print(equation)
```

```python
from sklearn.tree import DecisionTreeRegressor

# 1. Generate Complex Nonlinear Mockup Data (200 samples)
np.random.seed(42)

X = np.random.uniform(-3, 3, (200, 1))

y = (
    3 * np.sin(2 * X[:, 0])
    + 2 * np.cos(3 * X[:, 0])
    + 0.5 * X[:, 0]**2
)

y += np.random.normal(0, 0.8, 200)


# 2. Create and train Decision Tree Regression model
model = DecisionTreeRegressor(
    max_depth=5,
    random_state=42
)

model.fit(X, y)


# 3. Generate points for the regression curve
X_line = np.linspace(-3, 3, 300).reshape(-1, 1)
y_line = model.predict(X_line)


# 4. Visualize data and regression curve
plt.figure(figsize=(8, 5))

plt.scatter(
    X, y,
    alpha=0.6,
    label='Actual Data'
)

plt.plot(
    X_line,
    y_line,
    color='red',
    linewidth=2,
    label='Decision Tree Regression'
)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Decision Tree Regression (200 Samples)')
plt.legend()
plt.grid(True)
plt.show()


# 5. Display model information
print("Decision Tree Regression")
print("Maximum Depth:", model.get_depth())
print("Number of Leaves:", model.get_n_leaves())
```

Section 06 : Neural Network

<small>2 Dimension Model</small>

```python
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# 1. Generate Complex Nonlinear Mockup Data (300 samples)
np.random.seed(42)

X = np.random.uniform(-3, 3, (300, 1))

y = (
    3 * np.sin(2 * X[:, 0])
    + 2 * np.cos(3 * X[:, 0])
    + 0.5 * X[:, 0]**2
    + 2 * np.sin(2 * X[:, 0])
)

y += np.random.normal(0, 0.8, 300)


# 2. Create and train Neural Network Regression model
model = make_pipeline(
    StandardScaler(),
    MLPRegressor(
        hidden_layer_sizes=(64, 64),
        activation='relu',
        solver='adam',
        max_iter=3000,
        random_state=42,
        early_stopping=True
    )
)

model.fit(X, y)


# 3. Generate points for the regression curve
X_line = np.linspace(-3, 3, 300).reshape(-1, 1)
y_line = model.predict(X_line)


# 4. Visualize data and regression curve
plt.figure(figsize=(8, 5))

plt.scatter(
    X, y,
    alpha=0.6,
    label='Actual Data'
)

plt.plot(
    X_line,
    y_line,
    color='red',
    linewidth=2,
    label='Neural Network Regression'
)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Neural Network Regression (300 Samples)')
plt.legend()
plt.grid(True)
plt.show()


# 5. Display model information
mlp = model.named_steps['mlpregressor']

print("Neural Network Regression")
print("Hidden Layers:", mlp.hidden_layer_sizes)
print("Number of Iterations:", mlp.n_iter_)
print("Number of Input Features:", mlp.n_features_in_)
```

Section 07 : Different Epochs

```python
import plotly.graph_objects as go
import warnings
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.exceptions import ConvergenceWarning

# Ignore convergence warnings for this demonstration
warnings.filterwarnings("ignore", category=ConvergenceWarning)

# 1. Generate Complex Nonlinear Mockup Data (200 samples)
np.random.seed(42)

X1 = np.random.uniform(-3, 3, 200)
X2 = np.random.uniform(-3, 3, 200)

y = (
    1.5 * X1**3
    + 1.0 * X2**3
    - 5 * X1
    - 3 * X2
    + 5
)

y += np.random.normal(0, 3, 200)

# Combine X1 and X2
X = np.column_stack((X1, X2))


# 2. Generate points for the regression surface
X1_line = np.linspace(-3, 3, 80)
X2_line = np.linspace(-3, 3, 80)

X1_grid, X2_grid = np.meshgrid(X1_line, X2_line)

X_grid = np.column_stack((
    X1_grid.ravel(),
    X2_grid.ravel()
))


# 3. Compare different training iterations
iterations = [10, 50, 100, 500, 1000]

for n in iterations:

    # Create Neural Network model
    model = make_pipeline(
        StandardScaler(),
        MLPRegressor(
            hidden_layer_sizes=(100, 100),
            activation='relu',
            solver='adam',
            max_iter=n,
            early_stopping=False,
            random_state=42
        )
    )

    # Train model
    model.fit(X, y)

    # Predict regression surface
    y_grid = model.predict(X_grid)
    y_grid = y_grid.reshape(X1_grid.shape)

    # 4. Visualize data and regression surface
    fig = go.Figure()

    # Actual data points
    fig.add_trace(go.Scatter3d(
        x=X1,
        y=X2,
        z=y,
        mode='markers',
        marker=dict(
            size=3,
            color='blue',
            opacity=0.6
        ),
        name='Actual Data'
    ))

    # Neural Network regression surface
    fig.add_trace(go.Surface(
        x=X1_grid,
        y=X2_grid,
        z=y_grid,
        colorscale='Reds',
        opacity=0.8,
        name='Regression Surface'
    ))

    fig.update_layout(
        title=f'Neural Network Regression - {n} Iterations',
        scene=dict(
            xaxis_title='X1',
            yaxis_title='X2',
            zaxis_title='y'
        ),
        width=900,
        height=700
    )

    fig.show()

    # 5. Display model information
    mlp = model.named_steps['mlpregressor']

    print(f"\nTraining Iterations: {n}")
    print("Actual Iterations:", mlp.n_iter_)
    print("Hidden Layers:", mlp.hidden_layer_sizes)
```

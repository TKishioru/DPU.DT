Section A: Setup Environment

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error

import warnings
```

```python
# Setup Warning
warnings.filterwarnings('ignore')
```

---

Section B: Preprocessing Data

```python
# Load dataset
df_main = pd.read_csv('auto.csv')

# Drop unused columns
df_main = df_main.drop(['car_name'], axis=1)

# Drop records containing missing data
df_main.drop(df_main[df_main['horsepower']=='?'].index, inplace=True)

# Define datatype
df_main['mpg'] = df_main['mpg'].astype('float64')
df_main['cylinders'] = df_main['cylinders'].astype('float64')
df_main['displacement'] = df_main['displacement'].astype('float64')
df_main['horsepower'] = df_main['horsepower'].astype('float64')
df_main['weight'] = df_main['weight'].astype('float64')
df_main['acceleration'] = df_main['acceleration'].astype('float64')
df_main['model_year'] = df_main['model_year'].astype('float64')
df_main['origin'] = df_main['origin'].astype('float64')
df_main
```

```python
scalar = preprocessing.MinMaxScaler()

df_main[['mpg']] = scalar.fit_transform(df_main[['mpg']])
mpg_min = scalar.data_min_
mpg_max = scalar.data_max_

df_main[['cylinders']] = scalar.fit_transform(df_main[['cylinders']])
df_main[['displacement']] = scalar.fit_transform(df_main[['displacement']])
df_main[['horsepower']] = scalar.fit_transform(df_main[['horsepower']])
df_main[['weight']] = scalar.fit_transform(df_main[['weight']])
df_main[['acceleration']] = scalar.fit_transform(df_main[['acceleration']])
df_main[['model_year']] = scalar.fit_transform(df_main[['model_year']])
df_main[['origin']] = scalar.fit_transform(df_main[['origin']])
df_main 
```

```python
X = df_main.drop(['mpg'], axis=1)
y = df_main['mpg']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=45)
```

---

Section D: Testing Models

<small>1. Linear Regression</small>

```python
from sklearn.linear_model import LinearRegression

# Create regression model
regr = LinearRegression()
regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

<small>2. Non-Linear Regression</small>

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline

# Create nonlinear regression model
regr = make_pipeline(
    PolynomialFeatures(degree=2),
    LinearRegression()
)

regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)


print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

<small>3. Decision Tree Regression</small>

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Create regression model
regr = DecisionTreeRegressor(
    max_depth=5,
    random_state=42
)

regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)


print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

<small>4. Random Forest Regression</small>

```python
from sklearn.ensemble import RandomForestRegressor

# Create regression model
regr = RandomForestRegressor(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

<small>5. Multi Layer Perceptron</small>

```python
from sklearn.neural_network import MLPRegressor

# Create regression model
regr = MLPRegressor(
    hidden_layer_sizes=(128, 128),
    activation='relu',
    solver='adam',
    max_iter=5000,
    random_state=42
)

regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

<small>6. Support Vector Regression</small>

```python
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Create regression model
regr = SVR(
    kernel='rbf',
    C=100,
    epsilon=0.1
)

regr.fit(X_train, y_train)

# Evaluate regression model
y_test_pred = regr.predict(X_test)

# Inverse Scaling
y_test_pred_real = y_test_pred * (mpg_max - mpg_min) + mpg_min
y_test_real = y_test * (mpg_max - mpg_min) + mpg_min

# Calculate MAE and MSE
mae = mean_absolute_error(y_test_real, y_test_pred_real)
mse = mean_squared_error(y_test_real, y_test_pred_real)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
```

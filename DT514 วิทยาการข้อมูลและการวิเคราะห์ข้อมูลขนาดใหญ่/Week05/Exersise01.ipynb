```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read the ifood.xlsx and check the number of rows and columns in the dataset

df = pd.read_excel("../Week02/ifood.xlsx")
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}")
```

```python
# Calculate the mean, median, and standard deviation for the 'Income' column

print(f"Mean: {df['Income'].mean()}")
print(f"Median: {df['Income'].median()}")
print(f"Standard Deviation: {df['Income'].std()}")
```

```python
# Create a histogram of the "Age" column with gridlines and clear labels.

plt.hist(df['Age'], bins=20, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Age')
plt.grid(True)
plt.show()
```

```python
# Compute the range and interquartile range (IQR) of the "MntWines" column.

print(f"Range: {df['MntWines'].max() - df['MntWines'].min()}")
print(f"IQR: {df['MntWines'].quantile(0.75) - df['MntWines'].quantile(0.25)}")
```

```python
# Use "pd.cut" to divide "Income" into 4 ranges and count the number of customers in each group.
df['Income_Range'] = pd.cut(df['Income'], bins=4)
print(df['Income_Range'].value_counts())
```

```python
# Compute the average spending on meat products "MntMeatProducts" grouped by education level "Education".
df.groupby('Education')['MntMeatProducts'].mean()
```

```python
# Use GroupBy to calculate the average "Income" by marital status "Marital".
df.groupby('Marital')['Income'].mean()
```

```python
# Calculate the proportion of customers who responded to the latest campaign "Response = 1"
proportion = df[df['Response'] == 1].shape[0] / df.shape[0]
print(f"Proportion of customers who responded: {proportion}")
```

```python
# Create a bar chart showing the number of customers in each marital status category "Marital" using Matplotlib.
import matplotlib.pyplot as plt

df['Marital'].value_counts().plot(kind='bar')
plt.xlabel('Marital Status')
plt.ylabel('Number of Customers')
plt.title('Number of Customers by Marital Status')
plt.show()
```

```python
# Calculate the Pearson correlation between "Income" and "MntWines".
df['Income'].corr(df['MntWines'])
```

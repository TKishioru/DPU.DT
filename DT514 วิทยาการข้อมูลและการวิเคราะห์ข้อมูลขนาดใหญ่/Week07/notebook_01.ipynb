```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Example 01: Under Sampling 

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from imblearn.under_sampling import RandomUnderSampler

# 1. Create a synthetic 2D dataset
X, y = make_classification(
    n_samples=1000,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    weights=[0.90, 0.10],
    class_sep=1.0,
    random_state=42
)

# 2. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# 3. Apply Random Undersampling to training data only
rus = RandomUnderSampler(random_state=42)

X_train_under, y_train_under = rus.fit_resample(
    X_train, y_train
)

# 4. Display class distributions
print("Before Undersampling:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nAfter Undersampling:")
print(pd.Series(y_train_under).value_counts().sort_index())

# 5. Visualize the data before and after
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(
    X_train[:, 0], X_train[:, 1],
    c=y_train, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[0].set_title("Before Undersampling")
axes[0].set_xlabel("Feature 1")
axes[0].set_ylabel("Feature 2")

axes[1].scatter(
    X_train_under[:, 0], X_train_under[:, 1],
    c=y_train_under, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[1].set_title("After Random Undersampling")
axes[1].set_xlabel("Feature 1")
axes[1].set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay
)
from sklearn.pipeline import make_pipeline


# 1. Train SVM using the original training data
svm_original = make_pipeline(
    StandardScaler(),
    SVC(kernel="rbf", random_state=42)
)

svm_original.fit(X_train, y_train)

# 2. Train SVM using undersampled training data
svm_under = make_pipeline(
    StandardScaler(),
    SVC(kernel="rbf", random_state=42)
)

svm_under.fit(X_train_under, y_train_under)

# 3. Predict on the same test set
y_pred_original = svm_original.predict(X_test)
y_pred_under = svm_under.predict(X_test)

# 4. Compare performance
results = pd.DataFrame({
    "Original Data": [
        accuracy_score(y_test, y_pred_original),
        precision_score(y_test, y_pred_original, zero_division=0),
        recall_score(y_test, y_pred_original, zero_division=0),
        f1_score(y_test, y_pred_original, zero_division=0)
    ],
    "Undersampled Data": [
        accuracy_score(y_test, y_pred_under),
        precision_score(y_test, y_pred_under, zero_division=0),
        recall_score(y_test, y_pred_under, zero_division=0),
        f1_score(y_test, y_pred_under, zero_division=0)
    ]
}, index=["Accuracy", "Precision", "Recall", "F1-score"])

print(results.round(3))

# 5. Classification reports
print("\n=== Original Data ===")
print(classification_report(
    y_test, y_pred_original, zero_division=0
))

print("=== Undersampled Data ===")
print(classification_report(
    y_test, y_pred_under, zero_division=0
))

# 6. Confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_original,
    ax=axes[0], colorbar=False
)
axes[0].set_title("Original Data")

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_under,
    ax=axes[1], colorbar=False
)
axes[1].set_title("Undersampled Data")

plt.tight_layout()
plt.show()
```

Exampe 02: Over Sampling 

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import RandomOverSampler

# 1. Create a synthetic 2D dataset
X, y = make_classification(
    n_samples=1000,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    weights=[0.90, 0.10],
    class_sep=1.0,
    random_state=42
)

# 2. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# 3. Apply Random Oversampling to training data only
ros = RandomOverSampler(random_state=42)

X_train_over, y_train_over = ros.fit_resample(
    X_train, y_train
)

# 4. Display class distributions
print("Before Oversampling:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nAfter Random Oversampling:")
print(pd.Series(y_train_over).value_counts().sort_index())

# 5. Visualize the data before and after
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(
    X_train[:, 0], X_train[:, 1],
    c=y_train, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[0].set_title("Before Oversampling")
axes[0].set_xlabel("Feature 1")
axes[0].set_ylabel("Feature 2")

axes[1].scatter(
    X_train_over[:, 0], X_train_over[:, 1],
    c=y_train_over, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[1].set_title("After Random Oversampling")
axes[1].set_xlabel("Feature 1")
axes[1].set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay


# 6. Train SVM before Oversampling
svm_before = SVC(kernel="rbf", random_state=42)
svm_before.fit(X_train, y_train)

# 7. Train SVM after Oversampling
svm_after = SVC(kernel="rbf", random_state=42)
svm_after.fit(X_train_over, y_train_over)

# 8. Predict on the same test set
y_pred_before = svm_before.predict(X_test)
y_pred_after = svm_after.predict(X_test)

# 9. Compare classification performance
print("=== Before Oversampling ===")
print(classification_report(y_test, y_pred_before, zero_division=0))

print("=== After Random Oversampling ===")
print(classification_report(y_test, y_pred_after, zero_division=0))

# 10. Compare confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_before,
    ax=axes[0], cmap="Blues",
    colorbar=False
)
axes[0].set_title("SVM Before Oversampling")

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_after,
    ax=axes[1], cmap="Blues",
    colorbar=False
)
axes[1].set_title("SVM After Random Oversampling")

plt.tight_layout()
plt.show()
```

Example 03: SMOTE

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# 1. Create a synthetic 2D dataset
X, y = make_classification(
    n_samples=1000,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_clusters_per_class=1,
    weights=[0.90, 0.10],
    class_sep=1.0,
    random_state=42
)

# 2. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# 3. Apply SMOTE to training data only
smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train, y_train
)

# 4. Display class distributions
print("Before SMOTE:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nAfter SMOTE:")
print(pd.Series(y_train_smote).value_counts().sort_index())

# 5. Visualize the data before and after
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].scatter(
    X_train[:, 0], X_train[:, 1],
    c=y_train, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[0].set_title("Before SMOTE")
axes[0].set_xlabel("Feature 1")
axes[0].set_ylabel("Feature 2")

axes[1].scatter(
    X_train_smote[:, 0], X_train_smote[:, 1],
    c=y_train_smote, cmap="coolwarm",
    alpha=0.7, edgecolors="k", s=35
)
axes[1].set_title("After SMOTE")
axes[1].set_xlabel("Feature 1")
axes[1].set_ylabel("Feature 2")

plt.tight_layout()
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

# 6. Train SVM before SMOTE
svm_before = SVC(kernel="rbf", random_state=42)
svm_before.fit(X_train, y_train)

# 7. Train SVM after SMOTE
svm_after = SVC(kernel="rbf", random_state=42)
svm_after.fit(X_train_smote, y_train_smote)

# 8. Predict on the same test set
y_pred_before = svm_before.predict(X_test)
y_pred_after = svm_after.predict(X_test)

# 9. Compare classification performance
print("=== Before SMOTE ===")
print(classification_report(y_test, y_pred_before, zero_division=0))

print("=== After SMOTE ===")
print(classification_report(y_test, y_pred_after, zero_division=0))

# 10. Compare confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_before,
    ax=axes[0], cmap="Blues",
    colorbar=False
)
axes[0].set_title("SVM Before SMOTE")

ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred_after,
    ax=axes[1], cmap="Blues",
    colorbar=False
)

axes[1].set_title("SVM After SMOTE")

plt.tight_layout()
plt.show()
```

Example 04: ROC-PRC

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    roc_curve, roc_auc_score,
    precision_recall_curve, average_precision_score
)

from imblearn.under_sampling import RandomUnderSampler

# 1. Create a difficult imbalanced dataset
X, y = make_classification(
    n_samples=3000,
    n_features=10,
    n_informative=4,
    n_redundant=2,
    n_repeated=0,
    n_clusters_per_class=2,
    weights=[0.97, 0.03],
    class_sep=0.45,
    flip_y=0.02,
    random_state=42
)

print("Dataset shape:", X.shape)
print("\nOriginal class distribution:")
print(pd.Series(y).value_counts().sort_index())

# 2. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    stratify=y,
    random_state=42
)

# 3. Scale features using training data only
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Apply Random Undersampling to training data only
rus = RandomUnderSampler(random_state=42)

X_train_under, y_train_under = rus.fit_resample(
    X_train_scaled, y_train
)

print("\nBefore Undersampling:")
print(pd.Series(y_train).value_counts().sort_index())

print("\nAfter Random Undersampling:")
print(pd.Series(y_train_under).value_counts().sort_index())

# 5. Train SVM before Undersampling
svm_before = SVC(kernel="rbf", random_state=42)
svm_before.fit(X_train_scaled, y_train)

# 6. Train SVM after Undersampling
svm_after = SVC(kernel="rbf", random_state=42)
svm_after.fit(X_train_under, y_train_under)

# 7. Get decision scores on the same test set
score_before = svm_before.decision_function(X_test_scaled)
score_after = svm_after.decision_function(X_test_scaled)

# 8. Calculate ROC curves and AUC
fpr_before, tpr_before, _ = roc_curve(y_test, score_before)
fpr_after, tpr_after, _ = roc_curve(y_test, score_after)

auc_before = roc_auc_score(y_test, score_before)
auc_after = roc_auc_score(y_test, score_after)

print(f"\nROC-AUC Before Undersampling: {auc_before:.4f}")
print(f"ROC-AUC After Undersampling:  {auc_after:.4f}")

# 9. Calculate Precision-Recall curves and Average Precision
precision_before, recall_before, _ = precision_recall_curve(
    y_test, score_before
)
precision_after, recall_after, _ = precision_recall_curve(
    y_test, score_after
)

ap_before = average_precision_score(y_test, score_before)
ap_after = average_precision_score(y_test, score_after)

print(f"Average Precision Before Undersampling: {ap_before:.4f}")
print(f"Average Precision After Undersampling:  {ap_after:.4f}")

# 10. Plot ROC and Precision-Recall curves
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# ROC Curve
axes[0].plot(
    fpr_before, tpr_before,
    label=f"Before Undersampling (AUC = {auc_before:.3f})"
)
axes[0].plot(
    fpr_after, tpr_after,
    label=f"After Undersampling (AUC = {auc_after:.3f})"
)
axes[0].plot(
    [0, 1], [0, 1],
    linestyle="--", color="gray",
    label="Random Classifier"
)
axes[0].set_title("ROC Curve")
axes[0].set_xlabel("False Positive Rate")
axes[0].set_ylabel("True Positive Rate")
axes[0].legend()
axes[0].grid(alpha=0.3)

# Precision-Recall Curve
axes[1].plot(
    recall_before, precision_before,
    label=f"Before Undersampling (AP = {ap_before:.3f})"
)
axes[1].plot(
    recall_after, precision_after,
    label=f"After Undersampling (AP = {ap_after:.3f})"
)
axes[1].axhline(
    y=np.mean(y_test),
    linestyle="--", color="gray",
    label=f"Positive prevalence = {np.mean(y_test):.3f}"
)
axes[1].set_title("Precision-Recall Curve")
axes[1].set_xlabel("Recall")
axes[1].set_ylabel("Precision")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
